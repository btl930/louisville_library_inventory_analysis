```python
import pandas as pd
```

```python
lfpl_loc = pd.read_csv('../data/Louisville_KY_Free_Public_Libraries-2.csv')
```

```python
lfpl_loc.head()
```

```python
lfpl_loc[['city', 'state', 'zipcode']] = lfpl_loc['LFPL_LOC'].str.split (' ', expand=True)
```

```python
lfpl_loc = lfpl_loc.drop(['LFPL_LOC', 'LFPL_WEB', 'TELEPHONE', 'LFPL_ADD', 'city', 'state'], axis=1)
```

```python
lfpl_loc.head()
```

```python
lfpl_loc.rename(columns={'OBJECTID': 'library_id', 'LFPL_NAME': 'library_name', 'LATITUDE': 'latitude', 'LONGITUDE': 'longitude'}, inplace=True)
```
