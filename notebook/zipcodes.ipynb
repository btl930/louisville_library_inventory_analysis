```python
import pandas as pd
```

```python
zips = pd.read_csv('../data/Raw/Jefferson_County_KY_ZIP_Codes.csv')
```

```python
zips.head()
```

```python
zips = zips.drop('OBJECTID', axis=1)
```

```python
zips.head()
```

```python
zips.info()
```

```python
zips['ZIPCODE'] = zips['ZIPCODE'].astype(object)
```

```python
zips = zips.drop(['SHAPEAREA', 'SHAPELEN'], axis=1)
```

```python
zips = zips.rename(columns={'ZIPCODE':'zipcode'})
```

```python
zips.info()
```

```python
zips = zips.drop_duplicates(subset=['zipcode'])
```

```python
zips.to_csv('../data/Clean/clean_zips.csv', index=False)
```
