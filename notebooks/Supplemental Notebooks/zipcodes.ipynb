## Cleaning the Data

This dataset was fairly clean. Most work was done to allow it to work better with the sqlite database and to create better pin map visualizations.
- Dropped 'OBJECTID', 'SHAPEAREA', and 'SHAPELEN' columns as they were unnecessary for answering data questions.
- Renamed leftover columns to match syntax used in other notebooks and sqlite.
- Changed 'zipcode' column dtype to string to ensure data integrity when converted to sqlite.
- Dropped duplicate zipcodes.

```python
zips = zips.drop(['OBJECTID', 'SHAPEAREA', 'SHAPELEN'], axis=1)
```

```python
zips.head()
```

```python
zips.info()
```

```python
zips['ZIPCODE'] = zips['ZIPCODE'].astype(str)
```

```python
zips = zips.rename(columns={'ZIPCODE':'zipcode'})
```

```python
zips.info()
```

```python
zips = zips.drop_duplicates(subset=['zipcode'])
```

```python
zips['zipcode'].unique()
```

```python
zips.to_csv('../data/Clean/clean_zips.csv', index=False)
```
