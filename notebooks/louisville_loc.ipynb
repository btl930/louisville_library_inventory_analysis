## Cleaning the Data
This dataset was fairly clean. Most work was done to allow it to work better with the sqlite database. Visualizations using this data were done in the database notebook once the sqlite database was created.
- The column 'LFPL_LOC' was split so that zipcodes were isolated to their own column in order to be used as key.
- Dropped 'LFPL_LOC', 'LFPL_WEB', 'TELEPHONE', 'LFPL_ADD', 'city', and 'state' columns since they were unnecessary for answering the data questions.
- Renamed leftover columns to match syntax used in other notebooks and sqlite.
- Changed 'zipcode' column dtype to string to ensure data integrity when converted to sqlite.

```python
lfpl_loc[['city', 'state', 'zipcode']] = lfpl_loc['LFPL_LOC'].str.split (' ', expand=True)
```

```python
lfpl_loc = lfpl_loc.drop(['LFPL_LOC', 'LFPL_WEB', 'TELEPHONE', 'LFPL_ADD', 'city', 'state'], axis=1)
```

```python
lfpl_loc.rename(columns={'OBJECTID': 'library_id', 'LFPL_NAME': 'library_name', 'LATITUDE': 'latitude', 'LONGITUDE': 'longitude'}, inplace=True)
```

```python
lfpl_loc['zipcode'] = lfpl_loc['zipcode'].astype(str)
```
